# **Random Generation Function**

If you want to genrate some random numbers

It is a vector and values are between 0 and 1. It is a concept of normalization

```python
np.random.rand(10)
```

It is vector and values are between -3 to 3. It is a concept of Standardization. 
for that you need to use "np.random.randn()"

```python
np.random.randn(10)
```

If you want to genrate random integer then

```python
np.random.randint(8)
```

If you provide sinle value in randint(), it genrate single value between 1 to that number.

Now if you want to genrate multiple random numbers then
np.random.randint(low, high, size)

```python
np.random.randint(10,25,10)
```
